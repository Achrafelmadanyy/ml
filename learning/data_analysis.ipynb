```python
make_counts = df["make"].value_counts().sort_values()

plt.figure(figsize=(10, 12))
sns.barplot(x=make_counts.values, y=make_counts.index, color="steelblue")

plt.xlabel("Count")
plt.ylabel("Make")
plt.title("Number of Vehicles by Make")
plt.tight_layout()
plt.show()
```
